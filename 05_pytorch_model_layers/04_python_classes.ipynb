## Lesson 01: a class defines a kind of object

**Mathematical view:** [Related explanation and worked example](00_math_intuition.ipynb#model-state).

A **class** defines a kind of object: what data and behavior its objects can have. An **object** (also called an **instance**) is a particular object created from that class.

`class SimpleModel:` defines a class called `SimpleModel`. The indented body belongs to that class. `pass` is a placeholder that does nothing; it allows us to start with an empty class.

Calling `SimpleModel()` creates an object. Calling it again creates another object. Both objects are instances of the same class, but they are separate objects.

```python
# Define the class. Its body is empty for this first example.
class SimpleModel:
    pass

# Create two separate objects from the same class.
model_a = SimpleModel()
model_b = SimpleModel()

print("Class of model_a:", type(model_a).__name__)
print("Class of model_b:", type(model_b).__name__)
print("Are they the same object?", model_a is model_b)
```

## Lesson 02: store weight and bias with __init__ and self

**Mathematical view:** [Related explanation and worked example](00_math_intuition.ipynb#model-state).

Now define a class that stores a weight and a bias in each object.

- `def` defines a function. A function defined inside a class is a **method**.
- `__init__` is a special method Python calls to initialize a newly created instance. There are two underscores on each side of `init`.
- `self` refers to the particular instance that the method is working with. Python supplies it automatically when we create the object or call its method in the usual way.
- `self.weight` and `self.bias` are **attributes**: data stored on that particular object.

In `self.weight = weight`, the right-hand `weight` is the argument passed into initialization. The left-hand `self.weight` stores that value on the object. The same applies to bias.

`LinearModel(weight=3, bias=2)` creates an instance and initializes its attributes. We pass weight and bias; we do not pass `self` ourselves.

**Before running:** will the two objects have the same stored values just because they use the same class?

**My prediction:** _write here_


```python
# Define how each new object stores its initial weight and bias.
class LinearModel:
    def __init__(self, weight, bias):
        self.weight = weight  # Store the supplied weight on this object.
        self.bias = bias      # Store the supplied bias on this object.

# Each object gets the values supplied when that object is created.
model_a = LinearModel(weight=3, bias=2)
model_b = LinearModel(weight=1, bias=0)

print("Model A | weight:", model_a.weight, "bias:", model_a.bias)
print("Model B | weight:", model_b.weight, "bias:", model_b.bias)
```

## Lesson 03: give each object a prediction method

**Mathematical view:** [Related explanation and worked example](00_math_intuition.ipynb#model-state).

Add `predict(self, x)`, a method that uses the current object's weight and bias. `return` sends the calculated prediction back to the caller.

When we write `model_a.predict(2)`, Python supplies `model_a` as `self`, and the supplied `2` becomes `x`. Inside this call, `self.weight` means `model_a.weight`, and `self.bias` means `model_a.bias`.

The two objects use the same method code but their own stored values. This cell repeats the complete class definition with the new method and creates new instances of that updated definition.

**Before running:** calculate the prediction for input 2 from each model's weight and bias.

**My predictions:** _write here_


```python
# Keep initialization and the prediction calculation together in one class.
class LinearModel:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias

    def predict(self, x):
        # Use the stored values of the object on which predict was called.
        return self.weight * x + self.bias

# Create instances of the class definition that now includes predict.
model_a = LinearModel(weight=3, bias=2)
model_b = LinearModel(weight=1, bias=0)

print("Model A prediction for x=2:", model_a.predict(2))
print("Model B prediction for x=2:", model_b.predict(2))
```

## Lesson 04: change one object's data

**Mathematical view:** [Related explanation and worked example](00_math_intuition.ipynb#model-state).

The two instances have their own attributes. Assigning `model_a.weight = 4` changes model A's weight. It does not change model B's weight.

The prediction method reads the current values whenever we call it, so changing a stored weight changes that object's future predictions. This is a deliberate assignment of an ordinary Python number, not a gradient-based training update.

Reset both objects at the start of this cell so the experiment can be repeated.

**Before running:** after changing model A's weight, predict each model's output for input 2. What should stay the same?

**My predictions and explanation:** _write here_


```python
# Start from known values for a repeatable comparison.
model_a = LinearModel(weight=3, bias=2)
model_b = LinearModel(weight=1, bias=0)

# Change only the weight stored on model A.
model_a.weight = 4

print("Model A | weight:", model_a.weight, "bias:", model_a.bias)
print("Model B | weight:", model_b.weight, "bias:", model_b.bias)
print("Model A prediction for x=2:", model_a.predict(2))
print("Model B prediction for x=2:", model_b.predict(2))
```

## Vocabulary to revisit

| Term | Meaning in this example |
|---|---|
| Class | The definition named `LinearModel` |
| Object / instance | A particular model created with `LinearModel(...)` |
| Attribute | Stored data such as `model_a.weight` |
| Method | A function defined in the class, such as `predict` |
| `self` | The instance that the method is working with |
| `__init__` | The method that initializes the instance's attributes |
